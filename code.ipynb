```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import math

class ImageNoiseFilter:
    def __init__(self, image_path=None):
        self.original_image = np.array(Image.open(image_path).convert('L'))
        self.original_image = self.original_image.astype(np.float32) / 255.0
    
    def add_gaussian_noise(self, image, mean=0, std=0.1):
        noise = np.random.normal(mean, std, image.shape)
        noisy_image = image + noise
        return np.clip(noisy_image, 0, 1)
    
    def add_salt_pepper_noise(self, image, salt_prob=0.05, pepper_prob=0.05):

        noisy_image = image.copy()
        
        # Salt noise (white pixels)
        salt_coords = np.random.random(image.shape) < salt_prob
        noisy_image[salt_coords] = 1
        
        # Pepper noise (black pixels)
        pepper_coords = np.random.random(image.shape) < pepper_prob
        noisy_image[pepper_coords] = 0
        
        return noisy_image
    
    def add_speckle_noise(self, image, std=0.1):
        noise = np.random.normal(0, std, image.shape)
        noisy_image = image * (1 + noise)
        return np.clip(noisy_image, 0, 1)
    
    def manual_gaussian_filter(self, image, kernel_size=5, sigma=1.0):
        kernel = self.create_gaussian_kernel(kernel_size, sigma)
        filtered_image = self.convolution_2d(image, kernel)
        return filtered_image
    
    def create_gaussian_kernel(self, size, sigma):
        kernel = np.zeros((size, size))
        center = size // 2
        
        for i in range(size):
            for j in range(size):
                x, y = i - center, j - center
                kernel[i, j] = math.exp(-(x**2 + y**2) / (2 * sigma**2))

        kernel = kernel / np.sum(kernel) #normalisasi kernel
        return kernel
    
    def convolution_2d(self, image, kernel):
        kernel_size = kernel.shape[0]
        pad_size = kernel_size // 2

        padded_image = np.pad(image, pad_size, mode='reflect')

        result = np.zeros_like(image)

        for i in range(image.shape[0]):
            for j in range(image.shape[1]):
                region = padded_image[i:i+kernel_size, j:j+kernel_size]
                result[i, j] = np.sum(region * kernel)
        
        return result
    
    def manual_median_filter(self, image, kernel_size=5):
        pad_size = kernel_size // 2
        padded_image = np.pad(image, pad_size, mode='reflect')
        result = np.zeros_like(image)

        for i in range(image.shape[0]):
            for j in range(image.shape[1]):
                window = padded_image[i:i+kernel_size, j:j+kernel_size]
                result[i, j] = np.median(window)
        
        return result
    
    def calculate_psnr(self, original, processed):
        mse = np.mean((original - processed) ** 2)
        if mse == 0:
            return float('inf')
        
        max_pixel = 1.0  # Karena gambar dinormalisasi ke [0,1]
        psnr = 20 * math.log10(max_pixel / math.sqrt(mse))
        return psnr
    
    def calculate_mse(self, original, processed):
        return np.mean((original - processed) ** 2)
    
    def calculate_ssim(self, original, processed):
        mu1 = np.mean(original)
        mu2 = np.mean(processed)
        
        sigma1_sq = np.var(original)
        sigma2_sq = np.var(processed)
        sigma12 = np.mean((original - mu1) * (processed - mu2))
        
        c1 = 0.01 ** 2
        c2 = 0.03 ** 2
        
        ssim = ((2 * mu1 * mu2 + c1) * (2 * sigma12 + c2)) / \
               ((mu1**2 + mu2**2 + c1) * (sigma1_sq + sigma2_sq + c2))
        
        return ssim
    
    def process_and_compare(self):
        gaussian_noisy = self.add_gaussian_noise(self.original_image, std=0.15)
        salt_pepper_noisy = self.add_salt_pepper_noise(self.original_image, 0.05, 0.05)
        speckle_noisy = self.add_speckle_noise(self.original_image, std=0.2)

        sigmas = [0.5, 1.0, 2.0] 
        kernel_sizes = [3, 5, 7]  
        
        print("="*80)
        print("ANALISIS FILTERING DENGAN BERBAGAI PARAMETER")
        print("="*80)

        self.plot_original_and_noise(gaussian_noisy, salt_pepper_noisy, speckle_noisy)

        for sigma in sigmas:
            for kernel_size in kernel_sizes:
                print(f"\n{'='*60}")
                print(f"TESTING: Kernel Size = {kernel_size}x{kernel_size}, Sigma = {sigma}")
                print(f"{'='*60}")
                
                # Gaussian filtering dengan parameter saat ini
                gauss_filt_gaussian = self.manual_gaussian_filter(gaussian_noisy, kernel_size, sigma)
                gauss_filt_saltpepper = self.manual_gaussian_filter(salt_pepper_noisy, kernel_size, sigma)
                gauss_filt_speckle = self.manual_gaussian_filter(speckle_noisy, kernel_size, sigma)
                
                # Median filtering dengan kernel size saat ini
                med_filt_gaussian = self.manual_median_filter(gaussian_noisy, kernel_size)
                med_filt_saltpepper = self.manual_median_filter(salt_pepper_noisy, kernel_size)
                med_filt_speckle = self.manual_median_filter(speckle_noisy, kernel_size)
                
                # Menampilkan hasil untuk parameter ini
                self.plot_filtered_results(
                    gaussian_noisy, salt_pepper_noisy, speckle_noisy,
                    gauss_filt_gaussian, gauss_filt_saltpepper, gauss_filt_speckle,
                    med_filt_gaussian, med_filt_saltpepper, med_filt_speckle,
                    kernel_size, sigma
                )
                
                # Menghitung metrik untuk parameter ini
                self.calculate_parameter_metrics(
                    gaussian_noisy, salt_pepper_noisy, speckle_noisy,
                    gauss_filt_gaussian, gauss_filt_saltpepper, gauss_filt_speckle,
                    med_filt_gaussian, med_filt_saltpepper, med_filt_speckle,
                    kernel_size, sigma
                )

        self.comprehensive_analysis(gaussian_noisy, salt_pepper_noisy, speckle_noisy, sigmas, kernel_sizes)
    
    def plot_original_and_noise(self, gauss_noisy, sp_noisy, speckle_noisy):
        fig, axes = plt.subplots(1, 4, figsize=(16, 4))
        
        axes[0].imshow(self.original_image, cmap='gray', vmin=0, vmax=1)
        axes[0].set_title('Original Image')
        axes[0].axis('off')
        
        axes[1].imshow(gauss_noisy, cmap='gray', vmin=0, vmax=1)
        axes[1].set_title('Gaussian Noise')
        axes[1].axis('off')
        
        axes[2].imshow(sp_noisy, cmap='gray', vmin=0, vmax=1)
        axes[2].set_title('Salt & Pepper Noise')
        axes[2].axis('off')
        
        axes[3].imshow(speckle_noisy, cmap='gray', vmin=0, vmax=1)
        axes[3].set_title('Speckle Noise')
        axes[3].axis('off')
        
        plt.tight_layout()
        plt.show()
    
    def plot_filtered_results(self, gauss_noisy, sp_noisy, speckle_noisy,
                             gauss_filt_gauss, gauss_filt_sp, gauss_filt_speckle,
                             med_filt_gauss, med_filt_sp, med_filt_speckle,
                             kernel_size, sigma):
        fig, axes = plt.subplots(3, 3, figsize=(15, 12))
        
        # Baris 1: Noisy images
        axes[0,0].imshow(gauss_noisy, cmap='gray', vmin=0, vmax=1)
        axes[0,0].set_title('Gaussian Noise')
        axes[0,0].axis('off')
        
        axes[0,1].imshow(sp_noisy, cmap='gray', vmin=0, vmax=1)
        axes[0,1].set_title('Salt & Pepper Noise')
        axes[0,1].axis('off')
        
        axes[0,2].imshow(speckle_noisy, cmap='gray', vmin=0, vmax=1)
        axes[0,2].set_title('Speckle Noise')
        axes[0,2].axis('off')
        
        # Baris 2: Gaussian filtered
        axes[1,0].imshow(gauss_filt_gauss, cmap='gray', vmin=0, vmax=1)
        axes[1,0].set_title(f'Gaussian Filter\n(σ={sigma}, {kernel_size}x{kernel_size})')
        axes[1,0].axis('off')
        
        axes[1,1].imshow(gauss_filt_sp, cmap='gray', vmin=0, vmax=1)
        axes[1,1].set_title(f'Gaussian Filter\n(σ={sigma}, {kernel_size}x{kernel_size})')
        axes[1,1].axis('off')
        
        axes[1,2].imshow(gauss_filt_speckle, cmap='gray', vmin=0, vmax=1)
        axes[1,2].set_title(f'Gaussian Filter\n(σ={sigma}, {kernel_size}x{kernel_size})')
        axes[1,2].axis('off')
        
        # Baris 3: Median filtered
        axes[2,0].imshow(med_filt_gauss, cmap='gray', vmin=0, vmax=1)
        axes[2,0].set_title(f'Median Filter\n({kernel_size}x{kernel_size})')
        axes[2,0].axis('off')
        
        axes[2,1].imshow(med_filt_sp, cmap='gray', vmin=0, vmax=1)
        axes[2,1].set_title(f'Median Filter\n({kernel_size}x{kernel_size})')
        axes[2,1].axis('off')
        
        axes[2,2].imshow(med_filt_speckle, cmap='gray', vmin=0, vmax=1)
        axes[2,2].set_title(f'Median Filter\n({kernel_size}x{kernel_size})')
        axes[2,2].axis('off')
        
        fig.suptitle(f'Filtering Results: Kernel Size {kernel_size}x{kernel_size}, σ = {sigma}', 
                     fontsize=16, y=0.98)
        plt.tight_layout()
        plt.show()
    
    def calculate_parameter_metrics(self, gauss_noisy, sp_noisy, speckle_noisy,
                                   gauss_filt_gauss, gauss_filt_sp, gauss_filt_speckle,
                                   med_filt_gauss, med_filt_sp, med_filt_speckle,
                                   kernel_size, sigma):
        print(f"\nMETRIK UNTUK KERNEL SIZE {kernel_size}x{kernel_size}, SIGMA = {sigma}:")
        print("-" * 60)
        
        # Data untuk perhitungan
        noise_data = [
            ("Gaussian Noise", gauss_noisy, gauss_filt_gauss, med_filt_gauss),
            ("Salt & Pepper", sp_noisy, gauss_filt_sp, med_filt_sp),
            ("Speckle Noise", speckle_noisy, gauss_filt_speckle, med_filt_speckle)
        ]
        
        for noise_name, noisy_img, gauss_result, median_result in noise_data:
            # Metrik untuk Gaussian filter
            psnr_gauss = self.calculate_psnr(self.original_image, gauss_result)
            mse_gauss = self.calculate_mse(self.original_image, gauss_result)
            ssim_gauss = self.calculate_ssim(self.original_image, gauss_result)
            
            # Metrik untuk Median filter
            psnr_median = self.calculate_psnr(self.original_image, median_result)
            mse_median = self.calculate_mse(self.original_image, median_result)
            ssim_median = self.calculate_ssim(self.original_image, median_result)
            
            print(f"\n{noise_name}:")
            print(f"  Gaussian Filter - PSNR: {psnr_gauss:.2f} dB, MSE: {mse_gauss:.6f}, SSIM: {ssim_gauss:.4f}")
            print(f"  Median Filter   - PSNR: {psnr_median:.2f} dB, MSE: {mse_median:.6f}, SSIM: {ssim_median:.4f}")
            
            # Menentukan filter terbaik
            if psnr_gauss > psnr_median:
                print(f"  → Filter terbaik: Gaussian (selisih PSNR: +{psnr_gauss-psnr_median:.2f} dB)")
            else:
                print(f"  → Filter terbaik: Median (selisih PSNR: +{psnr_median-psnr_gauss:.2f} dB)")
    
    def comprehensive_analysis(self, gauss_noisy, sp_noisy, speckle_noisy, sigmas, kernel_sizes):
        print("\n" + "="*80)
        print("ANALISIS KOMPREHENSIF SEMUA PARAMETER")
        print("="*80)
        
        best_results = {
            'Gaussian Noise': {'filter': '', 'params': '', 'psnr': 0},
            'Salt & Pepper': {'filter': '', 'params': '', 'psnr': 0},
            'Speckle Noise': {'filter': '', 'params': '', 'psnr': 0}
        }
        
        noise_images = {
            'Gaussian Noise': gauss_noisy,
            'Salt & Pepper': sp_noisy,
            'Speckle Noise': speckle_noisy
        }
        
        print("\nTabel Perbandingan PSNR untuk semua kombinasi parameter:")
        print("-" * 80)

        print(f"{'Noise Type':<15} {'Filter':<8} {'Kernel':<6} {'Sigma':<5} {'PSNR (dB)':<10} {'MSE':<10} {'SSIM':<6}")
        print("-" * 80)
        
        for sigma in sigmas:
            for kernel_size in kernel_sizes:
                for noise_name, noise_img in noise_images.items():
                    # Gaussian filter
                    gauss_result = self.manual_gaussian_filter(noise_img, kernel_size, sigma)
                    psnr_gauss = self.calculate_psnr(self.original_image, gauss_result)
                    mse_gauss = self.calculate_mse(self.original_image, gauss_result)
                    ssim_gauss = self.calculate_ssim(self.original_image, gauss_result)
                    
                    # Median filter
                    median_result = self.manual_median_filter(noise_img, kernel_size)
                    psnr_median = self.calculate_psnr(self.original_image, median_result)
                    mse_median = self.calculate_mse(self.original_image, median_result)
                    ssim_median = self.calculate_ssim(self.original_image, median_result)

                    print(f"{noise_name:<15} {'Gaussian':<8} {kernel_size:<6} {sigma:<5} {psnr_gauss:<10.2f} {mse_gauss:<10.6f} {ssim_gauss:<6.4f}")
                    print(f"{'':<15} {'Median':<8} {kernel_size:<6} {'-':<5} {psnr_median:<10.2f} {mse_median:<10.6f} {ssim_median:<6.4f}")
                    
                    # Update hasil terbaik
                    if psnr_gauss > best_results[noise_name]['psnr']:
                        best_results[noise_name] = {
                            'filter': 'Gaussian',
                            'params': f'kernel={kernel_size}x{kernel_size}, σ={sigma}',
                            'psnr': psnr_gauss
                        }
                    
                    if psnr_median > best_results[noise_name]['psnr']:
                        best_results[noise_name] = {
                            'filter': 'Median',
                            'params': f'kernel={kernel_size}x{kernel_size}',
                            'psnr': psnr_median
                        }
                
                print("-" * 80)
        
        print("\nRINGKASAN HASIL TERBAIK:")
        print("="*50)
        for noise_name, result in best_results.items():
            print(f"{noise_name}:")
            print(f"  Filter terbaik: {result['filter']}")
            print(f"  Parameter: {result['params']}")
            print(f"  PSNR: {result['psnr']:.2f} dB")
            print()
        
        self.analyze_parameter_effects(gauss_noisy, sp_noisy, speckle_noisy, sigmas, kernel_sizes)
    
    def analyze_parameter_effects(self, gauss_noisy, sp_noisy, speckle_noisy, sigmas, kernel_sizes):
        print("ANALISIS PENGARUH PARAMETER:")
        print("="*40)
        
        noise_images = {
            'Gaussian Noise': gauss_noisy,
            'Salt & Pepper': sp_noisy,
            'Speckle Noise': speckle_noisy
        }
        
        print("\n1. PENGARUH SIGMA PADA GAUSSIAN FILTER:")
        print("-" * 40)
        for noise_name, noise_img in noise_images.items():
            print(f"\n{noise_name} (Kernel 5x5):")
            for sigma in sigmas:
                result = self.manual_gaussian_filter(noise_img, 5, sigma)
                psnr = self.calculate_psnr(self.original_image, result)
                print(f"  σ = {sigma}: PSNR = {psnr:.2f} dB")
        
        print("\n2. PENGARUH KERNEL SIZE:")
        print("-" * 40)
        for noise_name, noise_img in noise_images.items():
            print(f"\n{noise_name}:")
            print("  Gaussian Filter (σ=1.0):")
            for kernel_size in kernel_sizes:
                result = self.manual_gaussian_filter(noise_img, kernel_size, 1.0)
                psnr = self.calculate_psnr(self.original_image, result)
                print(f"    {kernel_size}x{kernel_size}: PSNR = {psnr:.2f} dB")
            
            print("  Median Filter:")
            for kernel_size in kernel_sizes:
                result = self.manual_median_filter(noise_img, kernel_size)
                psnr = self.calculate_psnr(self.original_image, result)
                print(f"    {kernel_size}x{kernel_size}: PSNR = {psnr:.2f} dB")

if __name__ == "__main__":
    processor = ImageNoiseFilter("images/kuda.jpg")
    processor.process_and_compare()
```
